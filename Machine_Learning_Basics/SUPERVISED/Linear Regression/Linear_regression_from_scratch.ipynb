##  Single Prediction element by element by loop

```python
x_vec = X_train[0,:]
print(x_vec)
print(x_vec.shape)
```

```python
def predict_single_loop(x,w,b):
    n=x.shape[0]
    p=0
    for i in range(n):
        p_i=x[i]*w[i]
        p=p+p_i
    p+=b
    return p
    
```

```python
x_vec = X_train[0,:]
second_house=X_train[2,:]
```

```python
print(predict_single_loop(x_vec,w_init,b_init))
```

# 5 Gradient Descent With Multiple Variables

```python
def gradient_descent(x, y, w, b, alpha, itrs):
    m = x.shape[0]
    w = w.copy()
    
    for i in range(itrs):
        predictions = np.dot(x, w) + b
        errors = predictions - y
        
        dj_dw = (1 / m) * (np.dot(x.T, errors))
        dj_db = (1 / m) * np.sum(errors)
        
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        
    # Un-indented! Returns only AFTER the loop finishes all 1000 iterations.
    return w, b

# ----------------------------------------
# Initialization and Execution
# ----------------------------------------
w_init = np.array([0., 0., 0., 0.])
b_init = 0.
iterations = 1000
alpha = 5.0e-7

# Assuming X_train and y_train are already defined in your environment
final_w, final_b = gradient_descent(X_train, y_train, w_init, b_init, alpha, iterations)

print(f"Optimized Weights: {final_w}")
print(f"Optimized Bias: {final_b}")
```
